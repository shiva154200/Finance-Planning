```python
import pandas as pd
```

```python
df=pd.read_csv('https://raw.githubusercontent.com/shiva154200/Datasets/refs/heads/main/finance_historical_trends_500.csv')
```

```python
df.head(10)
```

```python
df.info()
```

```python
assets = [
    "equity_return",
    "debt_return",
    "gold_return",
    "fd_return"
]

average_returns = df[assets].mean()

print(average_returns)
```

```python
average_returns = average_returns.sort_values(ascending=False)

print(average_returns)
```

```python
volatility = {
    "Equity": df["equity_volatility"].mean(),
    "Debt": df["debt_volatility"].mean(),
    "Gold": df["gold_volatility"].mean()
}

print(volatility)
```

```python
return_columns = {
    "Equity": "equity_return",
    "Debt": "debt_return",
    "Gold": "gold_return",
    "FD": "fd_return"
}
```

```python
summary = []

for asset, column in return_columns.items():

    row = {
        "Asset": asset,
        "Average Return": round(df[column].mean(), 2),
        "Minimum Return": round(df[column].min(), 2),
        "Maximum Return": round(df[column].max(), 2)
    }

    if asset != "FD":
        volatility_column = asset.lower() + "_volatility"
        row["Average Volatility"] = round(
            df[volatility_column].mean(), 2
        )
    else:
        row["Average Volatility"] = 0

    summary.append(row)

summary_df = pd.DataFrame(summary)

print(summary_df)
```

```python
best_asset = summary["Average Return"].idxmax()

print("Best average performer:", best_asset)
```

```python
most_stable = summary["Average Volatility"].idxmin()

print("Most stable asset:", most_stable)
```

```python
correlation_columns = [
    "equity_return",
    "debt_return",
    "gold_return",
    "fd_return",
    "inflation_rate",
    "interest_rate",
    "market_growth",
    "gdp_growth",
    "unemployment_rate"
]

correlation_matrix = df[correlation_columns].corr()

print("\nCorrelation Matrix:")
print(correlation_matrix.round(2))
```

```python
important_correlations = {
    "Equity vs Market Growth":
        df["equity_return"].corr(df["market_growth"]),

    "Equity vs Inflation":
        df["equity_return"].corr(df["inflation_rate"]),

    "Gold vs Inflation":
        df["gold_return"].corr(df["inflation_rate"]),

    "Debt vs Interest Rate":
        df["debt_return"].corr(df["interest_rate"]),

    "FD vs Interest Rate":
        df["fd_return"].corr(df["interest_rate"]),

    "Equity vs GDP Growth":
        df["equity_return"].corr(df["gdp_growth"])
}

for name, value in important_correlations.items():

    print(name, ":", round(value, 2))
```

```python
yearly_returns = df.groupby("year")[
    [
        "equity_return",
        "debt_return",
        "gold_return",
        "fd_return"
    ]
].mean()

print(yearly_returns)
```

```python
best_return_asset = summary_df.loc[
    summary_df["Average Return"].idxmax(),
    "Asset"
]

most_stable_asset = summary_df.loc[
    summary_df["Average Volatility"].idxmin(),
    "Asset"
]

print("Best average return:", best_return_asset)
print("Most stable investment:", most_stable_asset)
```

```python
condition_analysis = df.groupby("economic_condition")[
    [
        "equity_return",
        "debt_return",
        "gold_return",
        "fd_return"
    ]
].mean()
```

```python
analysis_result = {
    "summary": summary_df.to_dict("records"),

    "correlations": important_correlations,

    "best_return_asset": best_return_asset,

    "most_stable_asset": most_stable_asset,

    "yearly_returns": yearly_returns.to_dict(),

    "economic_conditions":
        condition_analysis.to_dict()
}
```
